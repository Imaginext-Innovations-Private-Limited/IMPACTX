```python
from pathlib import Path
import os
import pandas as pd
from PIL import Image
from collections import Counter

DATASET_PATH = Path(r"D:\FINAL DFDT\SELECTED_DATASET")

print("Dataset path:", DATASET_PATH)
print("Dataset exists:", DATASET_PATH.exists())
```

```python
IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png", ".bmp", ".webp"
}

if not DATASET_PATH.exists():
    raise FileNotFoundError(f"Dataset not found: {DATASET_PATH}")

print("MAIN FOLDERS:")
for folder in DATASET_PATH.iterdir():
    if folder.is_dir():
        print(" -", folder.name)

print("\nIMAGE COUNTS:")

for class_name in ["real", "fake"]:
    class_path = DATASET_PATH / class_name

    if not class_path.exists():
        print(f"{class_name.upper()}: FOLDER MISSING")
        continue

    images = [
        p for p in class_path.rglob("*")
        if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
    ]

    print(f"{class_name.upper()}: {len(images)} images")
```

```python
results = []

for class_name in ["real", "fake"]:
    class_path = DATASET_PATH / class_name

    if not class_path.exists():
        continue

    image_files = [
        p for p in class_path.rglob("*")
        if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
    ]

    corrupted = []

    for image_path in image_files:
        try:
            with Image.open(image_path) as img:
                img.verify()
        except Exception:
            corrupted.append(str(image_path))

    results.append({
        "Class": class_name.upper(),
        "Total Images": len(image_files),
        "Corrupted Images": len(corrupted),
        "Readable Images": len(image_files) - len(corrupted)
    })

    print(f"\n{class_name.upper()} corrupted files:")
    print(corrupted[:10])

inspection_df = pd.DataFrame(results)
display(inspection_df)
```
